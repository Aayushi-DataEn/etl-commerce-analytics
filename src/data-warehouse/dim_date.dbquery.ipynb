```sql
CREATE or replace table ecommerce_analytics.data_warehouse.dim_date AS
select  
  cast(date_format(d, 'yyyyMMdd') as INT) AS date_key ,
  d as full_date,
  day(d) as day_num,
  month(d) as month_num,
  year(d) as year_num ,
  quarter(d) as quarter_num
from (
  SELECT explode (
    sequence(
      to_date('2018-01-01'),
      to_date('2026-12-31'),
      interval 1 day
    )
  )  AS d
);

SELECT * from ecommerce_analytics.data_warehouse.dim_date;
```
