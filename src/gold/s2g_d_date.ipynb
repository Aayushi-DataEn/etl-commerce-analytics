```python
sales_orders_df = spark.read.table("ecommerce_analytics.silver.sales_orders")
sales_df = spark.read.table("ecommerce_analytics.silver.sales")
```

```python
from pyspark.sql.functions import *

dim_date = sales_orders_df.withColumn(
    "date",
    to_date(col("order_timestamp"))
).select("date").dropDuplicates()

```

```python
dim_date.display()
```

```python
dim_date = dim_date.withColumn(
    "date_key",
    date_format(col("date"), "yyyyMMdd").cast("int")
).withColumn(
    "day", dayofmonth("date")
).withColumn(
    "month", month("date")
).withColumn(
    "year", year("date")
).withColumn(
    "quarter", quarter("date")
).withColumn(
    "week", weekofyear("date")
)

```
