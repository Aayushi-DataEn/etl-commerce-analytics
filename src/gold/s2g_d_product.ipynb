```python
sales_df = spark.read.table("ecommerce_analytics.silver.sales")
```

```python
sales_df.display()
```

```python
from pyspark.sql.functions import col

dim_product = sales_df.select(
    col("product_id"),
    col("product_name"),
    col("unit"),
    col("product_category").alias("category")
).dropDuplicates(["product_id"])

dim_product.display()
```

```python
dim_product.write.mode("overwrite").saveAsTable("ecommerce_analytics.gold.dim_product")
```
